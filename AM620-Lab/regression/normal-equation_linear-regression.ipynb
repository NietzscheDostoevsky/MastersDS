```python
import numpy as np
```

```python
x = np.array([1,2,3,4,5]).reshape(-1,1)
y = np.array([1,2,3,4,5]).reshape(-1,1)

x0 = np.array([1,1,1,1,1]).reshape(-1,1)
```

```python
X = np.c_[x0,x]
Y = y
```

```python
print(X)
print(Y)
```

```python
theta = np.linalg.inv(X.T @ X) @ (X.T @ Y)
```

```python
print(np.round(theta,2))
```

```python
# using randomly generated data 

# Let the line be y = 3 + 7x

theta_true = [3, 7]
n = 100 # samples 
x = 2 * np.random.rand(n, 1)

y = theta_true[0] + theta_true[1] * x + np.random.randn(n, 1)
```

```python
x0 = np.ones(n).reshape(-1,1)

X = np.c_[x0, x]
Y = y
```

```python
print(X[:3])
print(Y[:3])
```

```python
theta = np.linalg.inv(X.T @ X) @ (X.T @ Y)
print(np.round(theta, 2))
```
