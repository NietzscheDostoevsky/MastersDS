```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(1, 20,20 )
y = 3 * x**2 + 5 + np.random.normal(0, 1, size=x.shape)
y_og = 3 * x**2 + 5
y[2] = 10
y[14] = 900

plt.scatter(x, y, alpha=0.5, color="blue")
plt.plot(x, y_og, color='red', label='Original Quadratic')
plt.title('Scatter Plot of Noisy Quadratic Data with outliers')
plt.xlim(1, 20)
plt.xticks(np.arange(1,21,1))

plt.scatter(x[2], y[2], color="red", s=100, label="Outlier")
plt.scatter(x[14], y[14], color="red", s=100)

plt.xlabel('x')
plt.ylabel('y')
plt.grid()
plt.show()
```
