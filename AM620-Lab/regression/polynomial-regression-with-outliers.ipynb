## Regression fit

```python
# Create design Matrix

def design_matrix(x, degree):
    
    X = np.ones((len(x), degree+1))
    for i in range(1,degree + 1, 1):
        X[:, i] = x ** i
    return X
```

```python
# Gradient Descent function for a given data matrix

def grad_descent(X, y, lr = 0.1, max_iters = 10000, tol = 1e-3):
    
    n,d = X.shape
    theta = np.zeros(d)
    Y = X.copy()
    # feature scaling 
    for col in range(1,d,1):
        feature = X[:,col]
        mu = np.mean(feature)
        sigma = np.std(feature)
        feature = (feature - mu) / sigma
        X[:, col] = feature 

    prev_loss = np.inf
    loss = np.inf
    y_pred = np.inf
    
    for i in range(max_iters):
        #forward step 
        y_pred = X @ theta 
        error = y_pred - y 
        loss = np.mean(error**2)

        # convergence check 
        if abs(prev_loss - loss) < tol:
            print(f"Convergence on step: {i+1}")
            break

        # backprop 

        grad = (2/n) * (X.T @ error)
        theta = theta - lr * grad
        prev_loss = loss

    # calculating R2 score 
    ss_res = np.sum((y-y_pred)**2)
    ss_tot = np.sum((y - np.mean(y))**2)

    r2 = 1 - (ss_res / ss_tot)
    
    return theta, loss, r2, y_pred



    
```

```python
degrees = [1,2,3,4,5]
plt.clf()
fig, axes = plt.subplots(len(degrees),1,figsize=(8,20))

for i, degree in enumerate(degrees):

    X = design_matrix(x, degree)
    theta, mse, r2, y_pred = grad_descent(X, y)
    
    print(f"--- Model: Polynomial Order {degree} ---")
    print(f"MSE: {mse:.2f}")
    print(f"R2 Score: {r2:.4f}")
    print(f"Trained Theta: {np.round(theta)}\n")
    
    ax = axes[i]
    ax.scatter(x, y, alpha = 0.5, color="blue", label="Noise data")
    ax.scatter([x[2], x[14]], [y[2], y[14]], color="red", s=100, label="Outliers")
    
    ax.plot(x, y_pred, color='green', linewidth=2, label=f"Degree {degree} Fit")
    ax.set_title(f"Polynomial Fit (Order {degree}) | MSE: {mse:.0f} | R2: {r2:.2f}")
    ax.legend()
    ax.grid()

plt.tight_layout()
plt.show()
```
