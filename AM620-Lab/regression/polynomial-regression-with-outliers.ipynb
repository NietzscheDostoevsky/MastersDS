## Regression fit

```python
# Create design Matrix

def design_matrix(x, degree):
    
    X = np.ones((len(x), degree+1))
    for i in range(1,degree + 1, 1):
        X[:, i] = x ** i
    return X
```

```python
# Gradient Descent function for a given data matrix

def grad_descent(X, y, lr = 0.1, max_iters = 10000, tol = 1e-3):
    
    n,d = X.shape
    theta = np.zeros(d)
    Y = X.copy()
    # feature scaling 
    for col in range(1,d,1):
        feature = X[:,col]
        mu = np.mean(feature)
        sigma = np.std(feature)
        feature = (feature - mu) / sigma
        X[:, col] = feature 

    prev_loss = np.inf

    for i in range(max_iters):
        #forward step 
        y_pred = X @ theta 
        error = y_pred - y 
        loss = np.mean(error**2)

        # convergence check 
        if abs(prev_loss - loss) < tol:
            print(f"Convergence on step: {i+1}")
            break

        # backprop 

        grad = (2/n) * (X.T @ error)
        theta = theta - lr * grad
        prev_loss = loss

    # calculating R2 score 
    ss_res = np.sum((y-y_pred)**2)
    ss_tot = np.sum((y - np.mean(y))**2)

    r2 = 1 - (ss_res / ss_tot)
    
    return Y, theta, loss, r2, i+1 



    
```
