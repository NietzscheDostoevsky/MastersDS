## Q1. Generate 1000 integers between 1 and 100, find mean and sd, count how many vals are above 50. 

```python
import numpy as np

arr = np.random.randint(1,101,size=1000)
print(f"Hightest value : {max(arr)}")
print(f"Lowest value: {min(arr)}")
print(f"Standard Deviation: {np.std(arr)}")
print(f"Mean: {np.mean(arr)}")
print(f"Median: {np.median(arr)}")

print(f"Values >50: {len([i for i in arr if i > 50])}")
print(f"value > 50: {np.sum(arr > 50)}")
```

## Q2. Write a functino that takes two numpy arrays of the same shape and returns a new array containing the element wise sum, difference, product and quotient

    a = np.array([10,20,30])
    b = np.array([1,2,3])

```python
def array_sum_diff_prod_quot(a,b):
    if len(a) != len(b):
        return AssertionError("both vectors should be of same length")
    
    sum = a + b
    diff = a - b
    prod = a * b
    quot = a / b
    
    return [sum, diff, prod, quot]


a = np.array([10,20,30])
b = np.array([1,2,3])

print(array_sum_diff_prod_quot(a,b))
```

## Q4. Write a function that finds the value in a numpy array that is closest to the given target number. 

    arr = np.array([10,22,37,41, 55])
    target = 40

```python
def find_closest(arr, target):
    closest = arr[0]
    min_diff = abs(closest - target)
    for i,num in enumerate(arr):
        diff = abs(num - target)
        if diff < min_diff:
            min_diff = diff
            closest = arr[i]
            print("new closest : " , closest)
    return closest
        


print(find_closest(arr = np.array([10,22,37,41, 55]), target=40))
```

## Q5. Given a 1D numpy array of temperature readings, replace any value that is more than 2SD away from the mean with the mean value of the array. 

    arr = np.array([22,23,21,24,100,22,23,-50,25])

Expected output: an array where outliers such as -50, 100 are replaced by the mean of the OG array

```python
arr = np.array([22,23,21,24,100,22,23,-50,25])

mean = np.mean(arr)
sd = np.std(arr)
sdx2 = sd * 2

for i, num in enumerate(arr):
    if abs(num - mean) > sdx2:
        arr[i] = mean

print(arr)
```

## Q8. Generate a 4x4 numpy array of random integers between 1 and 20, replace all elements that are divisible by 3 with their squares and set all elements less than 5 to 0. Return the modified array

```python
arr4x4 = np.random.randint(1, 21, size=16).reshape(4,4)
print(arr4x4)

for i in range(4):
    for j in range(4):
        element = arr4x4[i,j]
        if element % 3 == 0:
            arr4x4[i,j] = element ** 2
        if element < 5:
            arr4x4[i,j] = 0

print(arr4x4)
```

## Q9. Generate a 3x5 numpy array of floating point numbers, normalize each row such that the sum of elements in each row equals 1(ie, convert each row to a probability distribution). If a row sums to 0, leave it unchanged. 

```python
arr3x5 = np.random.rand(3,5)
print(arr3x5)

for i in range(3):
    rowsum = np.sum(arr3x5[i,:])
    if rowsum != 0:
        arr3x5[i,:] = arr3x5[i,:] / rowsum
print(arr3x5)
```

## Q10. Given two 4x4 numpy arrays A and B containing integers, compute the matrix product A, B. Then calculate the mean of all elements in the resulting matrix using : for each element in the product matrix, if it is greater than mean, multiply it by 3, othersise, divide it by 3. Return the modified matrix. 

```python
np.random.seed(37)

arr1 = np.random.randint(1,21,size=16).reshape(4,4)
arr2 = np.random.randint(1,21,size=16).reshape(4,4)

print(arr1)
print(arr2)

prod = np.matmul(arr1,arr2)
print("Prod" , prod)

mean = np.mean(prod)
print("Mean", mean)

modified = np.where(prod > mean , prod * 3, prod/3)
print(modified)
```
