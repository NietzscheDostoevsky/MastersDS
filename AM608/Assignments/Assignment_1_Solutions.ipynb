### Q1. Perform the following operations on a list:
**a.** Create a list of 5 names: 'Apple', 'Orange', 'Apricot', 'Mango', 'Pineapple'
**b.** Insert a name 'Guava' before 'Apricot'
**c.** Append a name 'Dragonfruit'
**d.** Delete 'Mango' from the list
**e.** Replace 'Apple' with 'Avocado'
**f.** Sort all names in the list
**g.** Print reversed sorted list

```python

fruits = ['Apple', 'Orange', 'Apricot', 'Mango', 'Pineapple']

fruits.insert(2, 'Guava')

fruits.append('Dragonfruit')

fruits.remove('Mango')

fruits[0] = 'Avocado'

fruits.sort()

fruits.reverse()

print(fruits)
```

### Q3. Write a program to find the mean, median and mode of a list of 10 numbers

```python
import statistics

numbers = [6, 7, 2, 7, 8, 2, 10, 7, 5, 9]
mean_val = statistics.mean(numbers)
print(f"Mean: {mean_val}")

# Calculate Median
median_val = statistics.median(numbers)
print(f"Median: {median_val}")

mode_val = statistics.mode(numbers)
print(f"mode val : {mode_val}")
```

### Q4. Suppose a list contains positive and negative numbers. Write a program to create two lists: one list containing all positive numbers (including 0) and another list containing all negative numbers.

```python
mixed_numbers = [10, -21, 4, -45, 66, -93, 0, 1]

positive_numbers = []
negative_numbers = []

for num in mixed_numbers:
    if num >= 0:
        positive_numbers.append(num)
    else:
        negative_numbers.append(num)

print(f"Original list: {mixed_numbers}")
print(f"Positive numbers list: {positive_numbers}")
print(f"Negative numbers list: {negative_numbers}")
```

### Q8. Create a list of tuples. Each tuple should contain a name and its marks in float. Write a program to sort the tuples in descending order by marks.

```python
student_marks = [('Saurabh', 98.5), ('Singh', 92.0), ('The', 76.5), ('Great', 92.0), ('Nietzsche', 85.5)]

# Sort using a lambda function as the key, targeting the second element (marks)
# Set reverse=True for descending order
sorted_students = sorted(student_marks, key= lambda x: x[1], reverse=True )

print(f"Original list: {student_marks}")
print(f"Sorted list by marks (descending): {sorted_students}")
```

### Q10. Write a program to create a set containing 10 randomly generated numbers in the range 15 to 45. Count how many of these numbers are less than 30. Delete all numbers which are greater than 35.

```python
import random

random_set = set()
while(len(random_set)<10):
    random_set.add(random.randint(15, 45))
    

# Count numbers less than 30
count_less_than_30 = [num for num in random_set if num < 30]
print(f"Numbers less than 30: {count_less_than_30}")


filtered_set = set([num for num in count_less_than_30 if num < 35])
print(f"Set after deleting numbers > 35: {filtered_set}")
```

### Q11. A set contains names which either begin either with A or with B. Write a program to separate out the names into two sets, one containing names beginning with A and another containing names beginning with B.

```python
names_set = {'Aaloo', 'Bandar', 'Akash', 'Bandariya', 'Alia', 'Bhatt'}

set_A = set()
set_B = set()

for name in names_set:
    if name[0] == 'A':
        set_A.add(name)
    else:
        set_B.add(name)

print(f"Original set: {names_set}")
print(f"Names starting with 'A': {set_A}")
print(f"Names starting with 'B': {set_B}")
```

### Q13. Suppose there are two dictionaries called names and registration containing names as key and year of admission as values. Write a program to merge the two dictionaries into a third dictionary.

```python
names1 = {'Saurabh': 2020, 'Singh': 2021}
registration2 = {'Virat': 2020, 'Kohli': 2022, 'Dhoni': 2023} 

merged_dict1 = names1.copy()
merged_dict1.update(registration2)

print(f"Dictionary 1: {names1}")
print(f"Dictionary 2: {registration2}")
print(f"Merged using update(): {merged_dict1}")


merged_dict2 = names1 | registration2
print(f"Merged using | operator: {merged_dict2}")
```

### Q15. Write a program to sort a dictionary in ascending order by key and descending order by value.

```python
data = {'Saurabh': 50, 'Singh': 20, 'Virat': 80, 'Kohli': 50}
print(f"Original dictionary: {data}")

# Sorting by key in ascending order
# sorted() on a dictionary's items returns a list of tuples

sorted_by_key = dict(sorted(data.items()))
print(f"Sorted by key (ascending): {sorted_by_key}")

# Sorting by value in descending order
sorted_by_value = dict(sorted(data.items(), key= lambda x:x[1], reverse=True))
print(f"Sorted by value (descending): {sorted_by_value}")
```
