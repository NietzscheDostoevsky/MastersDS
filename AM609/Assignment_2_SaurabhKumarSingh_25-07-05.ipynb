```python
# Solution 1

def sum_and_product() -> tuple[int,int]:
    num1 = int(input("Enter num 1: "))
    num2 = int(input("Enter num 2: "))
    num3 = int(input("Enter num 2: "))

    return (num1+num2+num3, num1*num2*num3)

print(sum_and_product())
```

```python
# Solution 2

def is_pangram(s: str) -> bool:

    alphabets = set("abcdefghijklmnopqrstuvwxyz")
    check = set(char.lower() for char in s if char.isalpha())
    if alphabets.issubset(check) :
        return True
    
    return False

print(is_pangram("The quick brown fox jumps over the lazy dog"))
print(is_pangram("Hello, world!"))
```

```python
# Solution 3

def change(s: str) -> str:
    a = s.split("-")
    a.sort()
    return "-".join(a)

print(change('this-is-python-lab'))  
# Output: 'is-lab-python-this'
```

```python
# Solution 4
def create_tuples() -> list:
   
    return [ (x, x**2, x**3) for x in range(1, 21)]

print(create_tuples())
```

```python
# Solution 5

def ispalindrome(s: str) -> bool :
    
    clean = "".join([c.lower() for c in s if c.isalpha()])
    
    #base case: 
    if len(clean) <= 1:
        return True
    if clean[0] != clean[-1]:
        return False
    return ispalindrome(clean[1:-1])

print(ispalindrome("level")) 
print(ispalindrome("A man a plan a canal Panama")) 
print(ispalindrome("hello")) 
```

```python
# Solution 6

def count_alphabets_digits(s:str) -> dict:
    count = {"alphabet":0, "digit":0}
    for c in s:
        if c.isalpha():
            count["alphabet"] += 1
        if c.isdigit():
            count["digit"] +=1 
    return count
    
    
print(count_alphabets_digits("Hello123World"))  # Output: {'alphabets': 10, 'digits': 3}
```

```python
# solution 7

def series(n: int) -> int:
    sum = 0
    for i in range(1, n+1):
        sum = sum + int(str(n) * i)
    return sum

print(series(5))  
print(series(9)) 
```

```python
# Solution 8
def remove_duplicates(s: list) -> list :
     return list(set(s))

print(remove_duplicates(["apple", "apple", "banana"]))
```

```python
# Solution 9

def calculate_exponent(a:int, b:int) -> int :
    
    #base case
    if b == 0:
        return 1
    elif b == 1:
        return a
    else: 
        return a * calculate_exponent(a, b-1)

print(calculate_exponent(2,3))
print(calculate_exponent(3,3))
```
