## Row-Echelon form
A matrix is in row-echelon form if it satisfies the two conditions: 
1. All rows consisting entirely of zeros are the the bottom of the matrix. 
2. The first non-zero number from the left of any non-zero row (called the ***pivot***) is to the right of the pivot of the row above it.

This structure creates a "staircase" pattern of leading non-zero entries. The goal of the algorithm is to use **Gaussian-Elimination** using **Elementary-Row Operations** to create zeros below each pivot.

### Example

$$
    \begin{bmatrix}
    1 & 2 & 3 \\
    2 & 3 & 5 \\
    3 & 7 & 6
    \end{bmatrix}

    \;\;\longrightarrow\;\;

    \begin{bmatrix}
    1 & 2 & 3 \\
    0 & -1 & -1 \\
    0 & 0 & -4
    \end{bmatrix}
$$




## First attempt it for a $3 \times 3$ matrix and later generalize the procedure

```python
# A test matrix who's result is already known
A = np.array([
    [1, 2, 3],
    [2, 3, 5],
    [3, 7, 6],
])
sp.pprint(sp.Matrix(A))
```

```python
B = np.array(A, dtype=np.float64) # commas separate the dimensions, and colons give the range.
B[1, :] = B[1, :] - (B[1, 0]/(B[0,0])) * (B[0,:])
B[2, :] = B[2, :] - (B[2, 0]/(B[0,0])) * (B[0,:])

B[2, :] = B[2, :] - (B[2, 1]/(B[1,1])) * (B[1,:])
assert B.shape == (3,3)
```

```python
# putting the above logic into a function which takes 3x3 array as an input
def reduce_row_echelon3x3(A):
    """Takes a 3x3 matrix as input, and reduces it into it's row echelon form"""
    B = np.array(A, dtype=np.float64)
    if B.shape != (3, 3):
        raise ValueError("Matrix must be 3x3!")

    #Reducing second row
    B[1, :] = B[1, :] - (B[1, 0]/(B[0,0])) * (B[0,:])
    B[2, :] = B[2, :] - (B[2, 0]/(B[0,0])) * (B[0,:])

    #Reducing third row
    B[2, :] = B[2, :] - (B[2, 1]/(B[1,1])) * (B[1,:])
    
    return B
```

```python
print(reduce_row_echelon3x3(A))
```
