## Row-Echelon form
A matrix is in row-echelon form if it satisfies the two conditions: 
1. All rows consisting entirely of zeros are the the bottom of the matrix. 
2. The first non-zero number from the left of any non-zero row (called the ***pivot***) is to the right of the pivot of the row above it.

This structure creates a "staircase" pattern of leading non-zero entries. The goal of the algorithm is to use **Gaussian-Elimination** using **Elementary-Row Operations** to create zeros below each pivot.

### Example

$$
    \begin{bmatrix}
    1 & 2 & 3 \\
    2 & 3 & 5 \\
    3 & 7 & 6
    \end{bmatrix}

    \;\;\longrightarrow\;\;

    \begin{bmatrix}
    1 & 2 & 3 \\
    0 & -1 & -1 \\
    0 & 0 & -4
    \end{bmatrix}
$$




## First attempt it for a $3 \times 3$ matrix and later generalize the procedure

```python
# A test matrix who's result is already known
A = np.array([
    [1, 2, 3],
    [2, 3, 5],
    [3, 7, 6],
])
sp.pprint(sp.Matrix(A))
```

```python
B = np.array(A, dtype=np.float64) # commas separate the dimensions, and colons give the range.
B[1, :] = B[1, :] - (B[1, 0]/(B[0,0])) * (B[0,:])
B[2, :] = B[2, :] - (B[2, 0]/(B[0,0])) * (B[0,:])

B[2, :] = B[2, :] - (B[2, 1]/(B[1,1])) * (B[1,:])
assert B.shape == (3,3)
```

```python
# putting the above logic into a function which takes 3x3 array as an input
def reduce_row_echelon3x3(A):
    """Takes a 3x3 matrix as input, and reduces it into it's row echelon form"""
    B = np.array(A, dtype=np.float64)
    if B.shape != (3, 3):
        raise ValueError("Matrix must be 3x3!")

    #Reducing second row
    B[1, :] = B[1, :] - (B[1, 0]/(B[0,0])) * (B[0,:])
    B[2, :] = B[2, :] - (B[2, 0]/(B[0,0])) * (B[0,:])

    #Reducing third row
    B[2, :] = B[2, :] - (B[2, 1]/(B[1,1])) * (B[1,:])
    
    return B
```

## Generalized Row Echelon

```python
from typing import Optional

# Defininng a tolerance for treating small numbers as zero, to avoid floating point rounding errors
EPSILON: float = 1.0e-8


def row_echelon(A: np.ndarray, epsilon: float = EPSILON) -> np.ndarray:
    """
    Given a mxn matrix, reduce it into it's row-echelon form using Gaussian Elimination.
    This is NOT an in place function. It returns a new Matrix, while maintaining the 
    original values inside the matrix provided. 
    
    Args:
        A (np.ndarray) : Input matrix
        epsilon (float) : small tolerance values for treating small values as zero.
    
    Return:
        np.ndarray: matrix in Row Echelon form     
    """
    
    B = np.array(A, dtype=np.float64) # deep copies the original array and converts the values into floating datatype for division.
    
    rows, cols = B.shape 
    pivot_row = 0 # The row index where current pivot element is located. 
                  # Start at the top row and moves downwards as elimination proceeds.
    
    for pivot_col in range(cols):
        # pivot_col : the column index where the pivott is chosen
        #            start at the leftmost column, and move rightwards
        #            if a column is all zeros below the current pivot row, 
        #               skip that column and move to the next column.
        
        # Step 1: Finding the pivot row
        max_row = None #the row index of the element with the largest absolute value in the current column(below the pivot row)
                       # after finding this row, swap it pivot row to maintain numerial stability, by maintaining the pivot as large as possible
        # finding the max row
        for r in range(pivot_row, rows):
            if abs(B[r, pivot_col]) > epsilon:
                max_row = r
                break
        
        if max_row is None:
            continue #means no pivot rows in this column, look in the next column. 
        
        # Step 2: Swap pivot row into position.
        if max_row != pivot_row:
            B[[pivot_row, max_row]] = B[[max_row, pivot_row]] # Standard row swap template in numpy
        
        # Step 3: Eliminate below pivot
        for r in range(pivot_row + 1, rows):
            if abs(B[r, pivot_col]) > epsilon: # only if that value is NOT effectively zero
                factor = B[r, pivot_col] / B[pivot_row, pivot_col]
                B[r, :] -= factor * B[pivot_row, :]
        
        pivot_row += 1
    
    # Step 4: Push zero rown to bottom
    # this block is just a technicality, it's not required as zero rows would have been automatically sunk to the bottom by now anyways. 
    nonzero_rows = [row for row in B if not np.allclose(row, 0, atol=epsilon)]
    zero_rows = [row for row in B if np.allclose(row, 0, atol=epsilon)]
    
    B = np.array(nonzero_rows + zero_rows)
    B[np.abs(B) < epsilon] = 0.0 # changes the almost zero entries to just 0
    return B
```
