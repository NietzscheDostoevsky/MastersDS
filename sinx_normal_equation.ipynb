```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Generate data
N = 100
x = np.linspace(-1, 1, N)
y = np.sin(np.pi * x)

# reshape for matrix ops
x = x.reshape(-1, 1)

# -----------------------
# Linear model: y = θ0 + θ1 x
# -----------------------
X_linear = np.hstack([np.ones((N, 1)), x])

theta_linear = np.linalg.inv(X_linear.T @ X_linear) @ X_linear.T @ y
y_hat_linear = X_linear @ theta_linear

# -----------------------
# Quadratic model: y = θ0 + θ1 x + θ2 x^2
# -----------------------
X_quad = np.hstack([np.ones((N, 1)), x, x**2])

theta_quad = np.linalg.inv(X_quad.T @ X_quad) @ X_quad.T @ y
y_hat_quad = X_quad @ theta_quad

# -----------------------
# Plot
# -----------------------
plt.figure(figsize=(8, 5))
plt.plot(x, y, label="True y = sin(πx)", linewidth=2)
plt.plot(x, y_hat_linear, label="Linear fit", linestyle="--")
plt.plot(x, y_hat_quad, label="Quadratic fit", linestyle="-.")
plt.legend()
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()

print("Linear theta:", theta_linear)
print("Quadratic theta:", theta_quad)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Data
N = 100
x = np.linspace(-1, 1, N)
y = np.sin(np.pi * x)

x_col = x.reshape(-1, 1)

def fit_polynomial(x, y, degree):
    """
    Fits y = sum_{k=0}^degree theta_k x^k using normal equations
    """
    X = np.hstack([x**k for k in range(degree + 1)])
    theta = np.linalg.inv(X.T @ X) @ X.T @ y
    return X @ theta, theta

degrees = [1, 2, 3, 5, 7, 9]

plt.figure(figsize=(9, 6))
plt.plot(x, y, label="True y = sin(πx)", linewidth=3)

for d in degrees:
    y_hat, theta = fit_polynomial(x_col, y, d)
    plt.plot(x, y_hat, label=f"Degree {d}")

plt.legend()
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Data
N = 100
x = np.linspace(-1, 1, N)
y = np.sin(np.pi * x)

x_col = x.reshape(-1, 1)

def fit_polynomial(x, y, degree):
    X = np.hstack([x**k for k in range(degree + 1)])
    theta = np.linalg.inv(X.T @ X) @ X.T @ y
    y_hat = X @ theta
    return y_hat, theta

degrees = [11]

plt.figure(figsize=(9, 6))
plt.plot(x, y, label="True y = sin(πx)", linewidth=3)

for d in degrees:
    y_hat, theta = fit_polynomial(x_col, y, d)
    plt.plot(x, y_hat, label=f"Degree {d}")

    print(f"\nDegree {d} polynomial:")
    for i, t in enumerate(theta):
        print(f"  theta_{i} = {t:.6f}")

plt.legend()
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```
