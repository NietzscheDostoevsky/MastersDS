```python
# End-to-end MNIST classification with multiple scikit-learn models
# ---------------------------------------------------------------
# Tasks:
# 1) Load data with fetch_openml
# 2) Split into train (first 60,000) and test (last 10,000)
# 3) Define four classifiers (default params; MLPClassifier with random_state=42)
# 4) Train, predict, and print accuracy, classification report, and 10x10 confusion matrix
#
# Note: Downloading MNIST from OpenML may take time the first run (it is cached locally).

import numpy as np
from sklearn.datasets import fetch_openml

# Classifiers
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.neural_network import MLPClassifier

# Metrics
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

def main():
    # ------------------------
    # Load the MNIST dataset
    # ------------------------
    # fetch_openml returns a Bunch with .data (features) and .target (labels).
    # as_frame=False yields NumPy arrays; parser='auto' chooses an efficient parser.
    mnist = fetch_openml(
        name="mnist_784",
        version=1,
        as_frame=False,
        parser="auto"
    )

    X = mnist.data              # Shape: (70000, 784)
    y = mnist.target            # Labels 0-9; may come as str or numeric depending on parser

    # Ensure labels are integers for metrics and confusion matrix indexing
    y = y.astype(int)

    # ---------------------------------------------------
    # Split into training (first 60k) and test (last 10k)
    # ---------------------------------------------------
    X_train, X_test = X[:60000], X[60000:]
    y_train, y_test = y[:60000], y[60000:]

    # --------------------------------
    # Define classifiers (default params)
    # --------------------------------
    models = {
        "KNeighborsClassifier": KNeighborsClassifier(),
        "DecisionTreeClassifier": DecisionTreeClassifier(),
        "GaussianNB": GaussianNB(),
        "MLPClassifier": MLPClassifier(random_state=42),  # default params with fixed seed
    }

    # Make NumPy printing more compact for the confusion matrix
    np.set_printoptions(linewidth=120, suppress=True)

    # -----------------------------------------
    # Train, predict, and print evaluation stats
    # -----------------------------------------
    for name, clf in models.items():
        print(f"\n--- Analyzing {name} ---")

        # Train the model
        clf.fit(X_train, y_train)

        # Predict on test set
        y_pred = clf.predict(X_test)

        # Accuracy score
        acc = accuracy_score(y_test, y_pred)
        print(f"Accuracy: {acc:.4f}")

        # Full classification report (precision, recall, F1-score per digit)
        print("\nClassification Report:")
        print(classification_report(y_test, y_pred, digits=4))

        # 10x10 confusion matrix (rows: true labels, columns: predicted labels)
        print("\nConfusion Matrix (10x10):")
        print(confusion_matrix(y_test, y_pred, labels=list(range(10))))

if __name__ == "__main__":
    main()
```
