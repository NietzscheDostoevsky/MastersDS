```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
dates = ['2025-01-01', '2025-01-02', '2025-01-03', '2025-01-06', '2025-01-07']
stock_a_price = [150.25, 151.50, 149.80, 155.00, 153.75]
volume = [150000, 120000, 180000, 250000, 160000]

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 6), sharex=True)

fig.suptitle('Stock Price Analysis')

ax1.plot(dates, stock_a_price, color='green', linestyle='-')
ax1.set_ylabel('Stock A Price')

ax2.bar(dates, volume, color='grey')
ax2.set_ylabel('Volume')
ax2.set_xlabel('Dates')

plt.xticks(rotation=45)
fig.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

```python
age = [25, 30, 45, 22, 55, 38, 41, 60]
annual_income = [45, 60, 100, 35, 120, 75, 85, 150]
purchase_score = [4, 7, 9, 3, 10, 6, 8, 9]
city_size = ['Small', 'Large', 'Large', 'Small', 'Large', 'Medium', 'Medium', 'Large']

df = pd.DataFrame({
    'Age': age,
    'Annual Income': annual_income,
    'Purchase Score': purchase_score,
    'City Size': city_size
})

plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x='Age',
    y='Annual Income',
    size='Purchase Score',
    hue='City Size',
    alpha=0.7,
    sizes=(50, 500),
    hue_order=['Small', 'Medium', 'Large']
)

plt.title('Customer Segmentation')
plt.xlabel('Age')
plt.ylabel('Annual Income (in thousands)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
region_x_sales = [100, 105, 110, 115, 120, 125, 130, 180]
region_y_sales = [90, 95, 100, 110, 115, 120, 125, 130, 140, 150]
data = [region_x_sales, region_y_sales]

plt.figure(figsize=(8, 6))

flierprops = dict(marker='D', markerfacecolor='red', markersize=8)
meanprops = dict(marker='o', markerfacecolor='cyan', markeredgecolor='black', markersize=8)

plt.boxplot(data, 
            labels=['Region X', 'Region Y'], 
            flierprops=flierprops, 
            showmeans=True, 
            meanprops=meanprops,
            patch_artist=True)

plt.title('Sales Distribution Comparison')
plt.ylabel('Sales')
plt.grid(axis='y')
plt.show()
```

```python
month = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
avg_temp = [15, 18, 22, 25, 28, 30]
total_rainfall = [60, 40, 30, 10, 5, 2]

fig, ax1 = plt.subplots(figsize=(10, 6))

color = 'tab:blue'
ax1.set_xlabel('Month')
ax1.set_ylabel('Average Temperature (C)', color=color)
line_plot = ax1.plot(month, avg_temp, color=color, marker='o', label='Temperature')
ax1.tick_params(axis='y', labelcolor=color)

ax2 = ax1.twinx()

color = 'tab:green'
ax2.set_ylabel('Total Rainfall (mm)', color=color)
bar_plot = ax2.bar(month, total_rainfall, color=color, alpha=0.6, label='Rainfall')
ax2.tick_params(axis='y', labelcolor=color)

plt.title('Monthly Weather Report')

lns = line_plot + [bar_plot]
labs = [l.get_label() for l in lns]
ax1.legend(lns, labs, loc='upper right')

fig.tight_layout()
plt.show()
```

```python
departments = ['Marketing', 'Sales', 'IT', 'HR']
female_employees = np.array([20, 15, 10, 5])
male_employees = np.array([10, 25, 30, 5])

totals = female_employees + male_employees

width = 0.5
fig, ax = plt.subplots(figsize=(10, 6))

ax.bar(departments, male_employees, width, label='Male', color='tab:blue')
ax.bar(departments, female_employees, width, 
       bottom=male_employees, label='Female', color='tab:pink')

ax.set_ylabel('Number of Employees')
ax.set_title('Employee Distribution by Department and Gender')
ax.legend()

for i in range(len(departments)):
    plt.text(i, totals[i] + 0.5, str(totals[i]), ha='center')

plt.show()
```
