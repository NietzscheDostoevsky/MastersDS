```python
a = [ [9, 9, 4, 8, 8, 1, 5, 3, 6, 3],
  [3, 2, 1, 9, 5, 1, 10, 7, 3],
  [5, 2, 6, 4, 5, 5, 4, 8, 2, 2],
  [8, 8, 3, 10, 10, 4, 3, 6, 9],
  [2, 1, 8, 7, 3, 1, 9, 5, 6, 2],
  [9, 2, 6, 5, 3, 9, 4, 6, 1, 10] ]
```

```python
results = []
for row in a:
    count = dict()
    for num in row:
        count[num] = count.get(num, 0) + 1
    #print(count)
    newrow = [0]*10
    for i in range(1, 11):
        newrow[i-1] = count.get(i,0)
    results.append(newrow)
for row in results:
    print(row)
    
```
