```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
months = ['Jan', 'Feb', 'March', 'Apr', 'May', 'Jun']
sales = [200, 250, 180, 300, 320, 280]

plt.plot(months, sales, 'bo-')

plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid(True)

max_sales = max(sales)
max_index = sales.index(max_sales)

plt.plot(months[max_index], sales[max_index], 'r*', markersize=15)
plt.show()
```

```python
fruits = ['Apple', 'Banana', 'Orange', 'Mango']
votes = [45, 30, 15, 10]
explode = (0.1, 0, 0, 0)

plt.pie(votes, labels=fruits, autopct='%1.1f%%', explode=explode)
plt.title('Favourite Fruits')
plt.axis('equal') 
plt.show()
```

```python
height = [150, 155, 160, 165, 170, 175, 180]
weight = [45, 50, 55, 58, 65, 70, 75]

plt.scatter(height, weight, color='green', marker='^')

plt.title('Height vs Weight')
plt.xlabel('Height (cm)')
plt.ylabel('Weight (kg)')
plt.show()
```

```python
students = ['Ali', 'Sara', 'Khan', 'Ayesha', 'Omeer']
marks = [85, 92, 78, 88, 90]

plt.bar(students, marks, color='orange')

plt.title('Student Marks')
plt.xlabel('Students')
plt.ylabel('Marks')
plt.grid(axis='y')

for i in range(len(students)):
    plt.text(i, marks[i] + 0.5, str(marks[i]), ha='center')

plt.show()
```

```python
scores = np.random.randint(40, 101, 50)

plt.hist(scores, bins=10, edgecolor='black')

plt.title('Exam Score Distribution')
plt.xlabel('Scores')
plt.ylabel('Frequency')
plt.show()
```

```python
cities = ['Lahore', 'Karachi', 'Islamabad', 'Faisalabad']
sales = [50000, 80000, 40000, 30000]

df = pd.DataFrame({
    'City': cities,
    'Sales': sales
})

df_sorted = df.sort_values(by='Sales')

plt.barh(df_sorted['City'], df_sorted['Sales'])
plt.title('Sales by City')
plt.xlabel('Sales')
plt.ylabel('City')

for index, value in enumerate(df_sorted['Sales']):
    plt.text(value, index, str(value))

plt.show()
```

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temp = [22, 24, 26, 25, 23, 27, 28]

plt.fill_between(days, temp, color='lightblue')
plt.plot(days, temp, color='darkblue')

plt.title('Weekly Temperature')
plt.xlabel('Day')
plt.ylabel('Temperature (°C)')
plt.show()
```

```python
class_a = [70, 75, 78, 80, 82, 85, 88]
class_b = [65, 68, 72, 75, 78, 80, 90]
data = [class_a, class_b]

bp = plt.boxplot(data, labels=['Class A', 'Class B'], patch_artist=True)

for box in bp['boxes']:
    box.set_facecolor('pink')

plt.title('Class Performance Comparison')
plt.ylabel('Scores')
plt.show()
```

```python
data = {
    'Maths': [72, 95, 61, 89, 54, 98, 76, 83],
    'Chemistry': [88, 58, 73, 96, 67, 81, 64, 90],
    'Physics': [65, 87, 80, 70, 93, 59, 88, 75],
    'History': [79, 92, 55, 84, 71, 90, 63, 68],
    'Arts': [91, 66, 78, 60, 85, 73, 97, 82]
}
index = ['Ali', 'Sara', 'Khan', 'Ayesha', 'Omar', 'Zara', 'Hassan', 'Noos']

df = pd.DataFrame(data, index=index)

plt.figure(figsize=(10, 7))
ax = sns.heatmap(df, annot=True, fmt='d', cmap='Blues', linewidths=0.5)
plt.title('Student Performance Across Subjects')

max_val = df.values.max()

for t in ax.texts:
    if float(t.get_text()) == max_val:
        t.set_weight('bold')
        t.set_color('white')

plt.show()
```
