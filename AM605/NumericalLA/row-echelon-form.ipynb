## Row-Echelon form
A matrix is in row-echelon form if it satisfies the two conditions:
1. All rows consisting entirely of zeros are the the bottom of the matrix.
2. The first non-zero number from the left of any non-zero row (called the ***pivot***) is to the right of the pivot of the row above it.

This structure creates a "staircase" pattern of leading non-zero entries. The goal of the algorithm is to use **Gaussian-Elimination** using **Elementary-Row Operations** to create zeros below each pivot.

### Example

$$
    \begin{bmatrix}
    1 & 2 & 3 \\
    2 & 3 & 5 \\
    3 & 7 & 6
    \end{bmatrix}

    \;\;\longrightarrow\;\;

    \begin{bmatrix}
    1 & 2 & 3 \\
    0 & -1 & -1 \\
    0 & 0 & -4
    \end{bmatrix}
$$
