### Smoke test on the environment

```python
import numpy as np
import scipy
import matplotlib
import matplotlib.pyplot as plt

# Testing numpy
print(f"NumPy version : {np.__version__}")
print(f"Scipy version : {scipy.__version__}")
print(f"Matplotlib version: {matplotlib.__version__}")
A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])
C = A @ B
print(f"A @ B : \n{C}")

P,L,U = scipy.linalg.lu(A)
print("SciPy LU decomposition of A:\n", f"P:\n{P}\nL:\n{L}\nU:\n{U}")

plt.figure()
plt.imshow(C)
plt.title("Heatmap of C")
plt.colorbar()
plt.show()
```
